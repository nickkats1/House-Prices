# 01 — Ingestion

Load the raw CSV, drop NaNs and duplicates, and persist a cleaned copy under `notebooks/_artifacts/` for the downstream steps to consume.

Production code path: `prices.data.ingestion.load_data`.

```python
import sys
from pathlib import Path

REPO_ROOT = Path.cwd().parents[1]
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

SCRATCH = REPO_ROOT / "notebooks" / "_artifacts"
SCRATCH.mkdir(parents=True, exist_ok=True)

import config
from prices.data.ingestion import load_data

print("data path:   ", config.DATA_PATH)
print("scratch dir: ", SCRATCH)
```

```python
df = load_data(file_path=config.DATA_PATH)
print(f"shape: {df.shape}")
df.head()
```

```python
df.dtypes
```

```python
missing = df.isna().sum()
print("missing values per column (post-cleaning):")
print(missing if missing.any() else "(none)")
print("\nduplicate rows (post-cleaning):", df.duplicated().sum())
```

```python
out_path = SCRATCH / "clean.parquet"
df.to_parquet(out_path, index=False)
print(f"wrote {len(df):,} rows -> {out_path.relative_to(REPO_ROOT)}")
```
